###`Q-1`: You are given a function definition. There might be several issues on execution of this function. You are asked to do exception handling for diffrent errors that this function goes in to `without altering this function`. And print error text.



Function parameters `l -> list, s -> could be anything`

```
def function(l: list, s, **args):
    last_element = l[-1]
    
    l[int(s)]=10
    any_element = l[int(s)+10]
    l[s]=10
    
    res = sum(l)
    
    p = args['p']
    # print(p)
    return res/last_element * p + any_element

```
Check for different function calls:-

```
function([1,2,1], 12)
function([1,2,1]*9, '1-2')
function([1,'2',1]*9, 12)
function([1,'2',1]*9, 12)
function([1,2,0]*9, 12  )
function([1,2,1]*9, 12, p=None)
function([1,2,0]*9, 12, p=10)
```

````python
def function(l: list, s, **args):
    try:
        last_element = l[-1]
        l[int(s)] = 10
        any_element = l[int(s) + 10]
        l[s] = 10
        res = sum(l)
        p = args['p']
        # print(p)
        return res / last_element * p + any_element
    except Exception as e:
        print("Error text:", e)


# --- Different Function Calls Check ---

# 1. Valid Call (agar p na ho toh KeyError aayega)
function([1, 2, 3, 4, 5], 2)

# 2. Invalid Index Call (IndexError)
function([1, 2], 0, p=5)

# 3. Invalid Data Type Call (ValueError / TypeError)
function("not a list", "abc", p=1)
````

###`Q-2:` You are given a code snippet. There might be several issues on execution of this code. You are asked to do exception handling for diffrent errors, condition is what ever happens we need to execute last line printing correct result of `sum of elements`.

List have elemnts as any no of  `key-pair dict with key as list index and value as any integer`, `integers` and `numeric-strings`. There is always only one element in the dict.


```
l = [{0:2},2,3,4,'5', {5:10}]
# For calculating sum of above list
s=0
for i in range(len(l)):
    #You can Edit code from here
    s += l[i].get(i)
    s += l[i]
    s += int(l[i])

```

````python
# Sample list input according to question
l = [{0: 2}, 2, 3, 4, '5', {5: 10}]


def calculate_sum(l):
    s = 0
    for i in range(len(l)):
        try:
            # 1. Agar element Dictionary hai (tulna/get check)
            if isinstance(l[i], dict):
                # Dict key-value pair ki value add karein
                s += l[i].get(i, list(l[i].values())[0])

            # 2. Agar element Integer ya Numeric String hai
            else:
                s += int(l[i])

        except Exception as e:
            print("Error text:", e)


    print("correct Result:", s)

    return s


# --- Function Execution ---
calculate_sum(l)
````

### `Q-4`: Number game program.

Write a number game program. Ask the user to enter a number. If the number is greater than number to be guessed, raise a **ValueTooLarge** exception. If the value is smaller the number to be guessed the, raise a **ValueTooSmall** exception and prompt the user to enter again. Quit the program only when the user enters the correct number. Also raise **GuessError** if user guess a number less than 1.

````python
# 1. Custom Exception Classes Definition
class GuessError(Exception):
    """User enters a number less than 1"""

    pass


class ValueTooLarge(Exception):
    """User guess is greater than target number"""

    pass


class ValueTooSmall(Exception):
    """User guess is smaller than target number"""

    pass


# 2. Main Game Logic
def number_game():
    target_number = 42  # Target number to guess

    while True:
        try:
            user_input = int(input("Enter a number: "))

            # Rule 1: Less than 1 check
            if user_input < 1:
                raise GuessError("Number must be 1 or greater!")

            # Rule 2: Too large check
            elif user_input > target_number:
                raise ValueTooLarge("This value is too large, try again.")

            # Rule 3: Too small check
            elif user_input < target_number:
                raise ValueTooSmall("This value is too small, try again.")

            # Rule 4: Correct Guess
            else:
                print("Congratulations! You guessed the correct number.")
                break

        except GuessError as e:
            print("GuessError:", e)
        except ValueTooLarge as e:
            print("ValueTooLarge:", e)
        except ValueTooSmall as e:
            print("ValueTooSmall:", e)
        except ValueError:
            print("Invalid Input! Please enter a valid integer.")


# Game start karne ke liye
# number_game()
````

````python
number_game()
````

### `Q-5:` Cast vote

Write a program that validate name and age as entered by the user to determine whether the person can cast vote or not. To handle the age, create **InvalidAge** exception and for name, create **InvalidName** exception. The name will be invalid when the string will be empty or name has only one word.

Example 1:

Input:

```bash
Enter the name:               goransh singh
Enter the age: 25
```

Output:

```bash
Goransh Singh  Congratulation !!! You can vote.
```

````python
# 1. Custom Exception Classes Define Karein
class InvalidAge(Exception):
    """Raised when age is less than 18"""

    pass


class InvalidName(Exception):
    """Raised when name contains more than one word"""

    pass


# 2. Main Function Logic
def cast_vote():
    try:
        name = input("Enter the name: ").strip()
        age = int(input("Enter the age: "))

        # Name Validation: Ek se zyada word hone par space milega
        if len(name.split()) > 1:
            raise InvalidName("Name should contain only one word.")

        # Age Validation: Vote dene ke liye age 18 ya usse zyada honi chahiye
        if age < 18:
            raise InvalidAge("Age must be at least 18 to vote.")

        # Title Case formatted output (e.g. goransh singh -> Goransh Singh)
        formatted_name = name.title()
        print(f"{formatted_name} Congratulation !!! You can vote.")

    except InvalidName as e:
        print("InvalidName Exception:", e)
    except InvalidAge as e:
        print("InvalidAge Exception:", e)
    except ValueError:
        print("ValueError: Please enter a valid numeric age.")


# Function Call
cast_vote()
````

### `Q-6`: Write a python function which infinitely prints natural numbers in a single line. Raise the **StopIteration** exception after displaying first 20 numnbers to exit from the program.

````python
def print_natural_number():
   n = 1
   try:
      while True:
          if n > 20:
              raise StopIteration("Reached 20 numbers, stopping loop.")

          print(n, end= " ")
          n += 1
   except StopIteration as e:
    print(f"\n{e}")


print_natural_number()
````
