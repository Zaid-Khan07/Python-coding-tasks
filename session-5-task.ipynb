###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

res = {}

for sub in test_list:
    res.setdefault(sub[0], []).extend(sub[1:])

output = [tuple([key] + val) for key, val in res.items()]

print(output)
````

###`Q2:` Multiply Adjacent elements (both side) and take sum of right and lest side multiplication result.


For eg.
```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
test_tuple = (1, 5, 7, 8, 10)
n = len(test_tuple)

res = []
for i in range(n):
    if i == 0:
        # First element: sirf right side se multiply
        val = test_tuple[i] * test_tuple[i + 1]
    elif i == n - 1:
        # Last element: sirf left side se multiply
        val = test_tuple[i] * test_tuple[i - 1]
    else:
        # Middle elements: (Left side multiply) + (Right side multiply)
        val = (test_tuple[i] * test_tuple[i - 1]) + (test_tuple[i] * test_tuple[i + 1])
    res.append(val)

output = tuple(res)
print(output)
````

###`Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
list1 = [{'hi', 'bye'},{'Geeks', 'forGreels'},('a', 'b'),['hi', 'bye'],['a', 'b']]

list_count = 0
set_count = 0
tuple_count = 0

for item in list1:
    if isinstance(item, list):
        list_count += 1
    elif isinstance(item, set):
        set_count += 1
    elif isinstance(item, tuple):
        tuple_count += 1

print("List-", list_count)
print("set-", list_count)
print("Tuple-", tuple_count)
````

###`Q1:` Write a program to find set of common elements in three lists using sets.
```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```

````python
ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

# Lists ko sets mein convert karke intersection (& operator) find karna
common = set(ar1) & set(ar2) & set(ar3)

# Output ko list format mein convert karna
output = list(common)

print(output)
````

###`Q2:` Write a program to count unique number of vowels using sets in a given string. Lowercase and upercase vowels will be taken as different.

`Input:`
```
Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
```
`Output:`
```
No of unique vowels-6
```

````python
str1 = "hands-on data science mentorship program with live classes at affordable fee only on compusx"
vowels = set("aeiouAEIOU")

unique_vowels = set()

for char in str1:
    if char in vowels:
        unique_vowels.add(char)

no_of_unique_vowels = len(unique_vowels)

print("No of unique vowels-", no_of_unique_vowels)
````

### `Q2`: Replace words from Dictionary. Given String, replace it’s words from lookup dictionary.

**Example 1:**

Input:

```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
```

Output:

```bash
CampusX is the best channel for Data-Science students.
```

**Example 2:**

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:
```bash
CampusX best for DS students.
```

````python
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}

# String ko words mein split karke list banayein
words = test_str.split()

# Har word ko check karke lookup dictionary se replace karein
res = []
for word in words:
    # Agar word ke aage punctuation marks (.) hain toh unhe handle karne ke liye dict check
    res.append(repl_dict.get(word, word))

# Wapas string mein join kar dein
output = " ".join(res)

print(output)
````

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]

n = len(key_list)
res = []

# Loop 0 se start karke key_list ki length ke step size se aage badhega
for i in range(0, len(test_list), n):
    # test_list se utne elements ki sublist nikalo jitni keys hain
    sub_values = test_list[i : i + n]

    # keys aur values ko pair karke dictionary banao
    d = dict(zip(key_list, sub_values))

    res.append(d)

print(res)
````
